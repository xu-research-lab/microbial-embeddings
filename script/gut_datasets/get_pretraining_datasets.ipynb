```python
import biom
import numpy as np
import pandas as pd
```

```python
### import data
gut_df = biom.load_table("../../data/table_gut_all.biom")
meta_disease = pd.read_csv("../../data/metadata_disease_classification.tsv", sep = "\t")
```

```python
meta_disease.head()
```

```python
sid_all = gut_df.ids(axis="sample")
sid_disease = meta_disease["sample"].values
```

```python
sid_remove_disease = sid_all[[i not in sid_disease for i in sid_all]]
```

```python
gut_df_remove_disease = gut_df.filter(sid_remove_disease, axis="sample", inplace=False)
gut_df_remove_disease.remove_empty()
```

```python
with biom.util.biom_open('../../data/gut_pretraining.biom', 'w') as f:
    gut_df_remove_disease.to_hdf5(f, '')
```
